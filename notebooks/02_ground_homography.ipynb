# 02 - Ground-Plane Homography

Use one manually captured empty-scene image with 4 or more measured ground markers, preferably 6 to 10. Enter marker pixels and measured ground coordinates in `data/ground_markers.csv` with columns `image_x,image_y,ground_x_m,ground_y_m`.

```python
from pathlib import Path

import numpy as np

from pedflow.calibration import compute_ground_homography, merge_and_save_calibration, read_marker_csv
from pedflow.geometry import load_calibration

INTRINSICS_PATH = Path("../outputs/calibration_intrinsics.json")
MARKERS_CSV = Path("../data/ground_markers.csv")
OUTPUT_PATH = Path("../outputs/calibration.json")
```

Optional marker clicking workflow: open the empty-scene calibration image in any image viewer or annotation tool, record each marker pixel coordinate, then type the measured meter coordinates into `data/ground_markers.csv`. This keeps capture manual and avoids any field image stream or automatic capture mode.

```python
intrinsics = load_calibration(INTRINSICS_PATH)
markers = read_marker_csv(MARKERS_CSV)
markers
```

```python
homography = compute_ground_homography(
    marker_points=markers,
    camera_matrix=np.asarray(intrinsics["K"], dtype=float),
    dist_coeffs=np.asarray(intrinsics["dist"], dtype=float),
)

calibration = merge_and_save_calibration(intrinsics, homography, OUTPUT_PATH)
print(f"Saved {OUTPUT_PATH}")
print("Mean marker residual (m):", calibration["ground_marker_mean_residual_m"])
print("Max marker residual (m):", calibration["ground_marker_max_residual_m"])
```
