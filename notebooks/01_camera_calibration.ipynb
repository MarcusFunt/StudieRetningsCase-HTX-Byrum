# 01 - Camera Intrinsic Calibration

Use manually captured ChArUco-board images from the mounted Grove Vision AI V2 camera. These images are calibration-only inputs and should not contain pedestrians. Delete them after `calibration_intrinsics.json` has been verified.

Generate the printable board and metadata first:

```powershell
python ../scripts/generate_charuco_board.py --output-dir ../outputs/charuco_board
```

Print the generated PDF at 100% scale.

````python
from pathlib import Path

from pedflow.calibration import calibrate_camera_from_charuco
from pedflow.geometry import save_calibration

IMAGE_DIR = Path("../data/calibration_images/charuco")
BOARD_METADATA = Path("../outputs/charuco_board/charuco_board.json")
OUTPUT_PATH = Path("../outputs/calibration_intrinsics.json")
MIN_CHARUCO_CORNERS = 8
````

````python
image_paths = sorted(
    list(IMAGE_DIR.glob("*.jpg"))
    + list(IMAGE_DIR.glob("*.jpeg"))
    + list(IMAGE_DIR.glob("*.png"))
)
print(f"Found {len(image_paths)} calibration images")
image_paths[:5]
````

````python
intrinsics = calibrate_camera_from_charuco(
    image_paths=image_paths,
    board_metadata_path=BOARD_METADATA,
    min_corners=MIN_CHARUCO_CORNERS,
)
save_calibration(intrinsics, OUTPUT_PATH)
print(f"Saved {OUTPUT_PATH}")
print("RMS reprojection error (px):", intrinsics["rms_reprojection_error_px"])
print("Usable images:", len(intrinsics["used_images"]))
print("Detected ChArUco corners per image:", intrinsics["detected_charuco_corners_per_image"])
````
