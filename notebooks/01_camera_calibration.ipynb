# 01 - Camera Intrinsic Calibration

Use manually captured checkerboard images from the mounted Grove Vision AI V2 camera. These images are calibration-only inputs and should not contain pedestrians. Delete them after `calibration_intrinsics.json` has been verified.

```python
from pathlib import Path

from pedflow.calibration import calibrate_camera_from_checkerboard
from pedflow.geometry import save_calibration

IMAGE_DIR = Path("../data/calibration_images/checkerboard")
OUTPUT_PATH = Path("../outputs/calibration_intrinsics.json")

# Inner corners, not printed squares. A 10 x 7 square board has 9 x 6 inner corners.
PATTERN_SIZE = (9, 6)
SQUARE_SIZE_M = 0.025
```

```python
image_paths = sorted(
    list(IMAGE_DIR.glob("*.jpg"))
    + list(IMAGE_DIR.glob("*.jpeg"))
    + list(IMAGE_DIR.glob("*.png"))
)
print(f"Found {len(image_paths)} calibration images")
image_paths[:5]
```

```python
intrinsics = calibrate_camera_from_checkerboard(
    image_paths=image_paths,
    pattern_size=PATTERN_SIZE,
    square_size_m=SQUARE_SIZE_M,
)
save_calibration(intrinsics, OUTPUT_PATH)
print(f"Saved {OUTPUT_PATH}")
print("RMS reprojection error (px):", intrinsics["rms_reprojection_error_px"])
print("Usable images:", len(intrinsics["used_images"]))
```
